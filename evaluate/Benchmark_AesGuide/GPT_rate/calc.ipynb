```python
# Completeness:
import json
from collections import Counter
def load_json(file_path):
    with open(file_path, 'r') as f:
        return json.load(f)

Completeness_file = "Venus_CVPR2026/evaluate/Benchmark_AesGuide/GPT_rate/Completeness.json"
data = load_json(Completeness_file)

scores = []
for key, value in data.items():
    score = float(value.split(":")[1].strip())
    scores.append(score)

score_counts = Counter(scores)
total_scores = sum(scores)
num_scores = len(scores)
average_score = total_scores / num_scores if num_scores > 0 else 0

print("Score Counts:", dict(score_counts))
print(f"Average Score: {average_score:.2f}")
```
